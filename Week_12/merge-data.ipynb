```python
"""
=============================================================================
MERGE 20 NĂM — Upload tất cả output NB1 vào 1 Kaggle Dataset
=============================================================================
CÁCH DÙNG:
  1. Tạo Kaggle Dataset mới (VD: "era5-20years-output")
  2. Upload TẤT CẢ các file vào: climate_weekly_*.parquet, flunet_clean_*.parquet
     (Mỗi năm có 1 climate + 1 flunet, tổng ~40 files)
  3. Tạo Notebook mới, add dataset, paste code này, chạy
  4. Download output: climate_weekly.parquet + flunet_clean.parquet
  5. Upload 2 file này làm dataset mới cho NB2, NB3
=============================================================================
"""

import glob, os
import pandas as pd

# Tìm tất cả parquet
ALL = sorted(glob.glob('/kaggle/input/**/*.parquet', recursive=True))
climate_files = [f for f in ALL if 'climate_weekly' in os.path.basename(f)]
flu_files     = [f for f in ALL if 'flunet_clean'   in os.path.basename(f)]

print(f'Climate files: {len(climate_files)}')
print(f'FluNet files:  {len(flu_files)}')

# ── Merge Climate ─────────────────────────────────────────────────
dfs = []
for f in sorted(climate_files):
    df = pd.read_parquet(f)
    print(f'  {os.path.basename(f):40s} {len(df):>6,} rows  years={sorted(df["year"].unique())}')
    dfs.append(df)

climate = pd.concat(dfs, ignore_index=True)
climate = climate.drop_duplicates(subset=['country_iso3','year','week'])
climate = climate.sort_values(['country_iso3','year','week']).reset_index(drop=True)
climate.to_parquet('/kaggle/working/climate_weekly.parquet', index=False)
print(f'\n✅ climate_weekly.parquet: {len(climate):,} rows | {sorted(climate["year"].unique())}')

# ── Merge FluNet ──────────────────────────────────────────────────
dfs = []
for f in sorted(flu_files):
    df = pd.read_parquet(f)
    print(f'  {os.path.basename(f):40s} {len(df):>6,} rows')
    dfs.append(df)

flu = pd.concat(dfs, ignore_index=True)
flu = flu.drop_duplicates(subset=['country_iso3','year','week'])
flu = flu.sort_values(['country_iso3','year','week']).reset_index(drop=True)
flu.to_parquet('/kaggle/working/flunet_clean.parquet', index=False)
print(f'\n✅ flunet_clean.parquet: {len(flu):,} rows | {sorted(flu["year"].unique())}')

print(f'\n🎉 Merge {len(climate_files)} năm hoàn tất')
```
